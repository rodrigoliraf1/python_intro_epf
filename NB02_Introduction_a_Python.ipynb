## 4. Comparaisons : poser une question à Python
Une **comparaison** est une expression qui compare deux valeurs et produit un **booléen** : `True` (vrai) ou `False` (faux). Elle ne choisit pas à elle seule une action ; elle fournit le résultat logique dont une condition pourra se servir.

| Opérateur | Question posée | Exemple vrai |
|---|---|---|
| `==` | Les valeurs sont-elles égales ? | `5 == 5` |
| `!=` | Les valeurs sont-elles différentes ? | `5 != 8` |
| `<` | Strictement inférieur ? | `5 < 8` |
| `<=` | Inférieur ou égal ? | `5 <= 5` |
| `>` | Strictement supérieur ? | `8 > 5` |
| `>=` | Supérieur ou égal ? | `8 >= 8` |

**Attention aux frontières :** « au moins 20 » inclut 20 (`>= 20`) ; « plus de 20 » l'exclut (`> 20`). Comparez des grandeurs compatibles : deux énergies dans la même unité, par exemple.

````python
energie_restante_wh = 20
reserve_minimale_wh = 20
reserve_suffisante = energie_restante_wh >= reserve_minimale_wh
print(reserve_suffisante)
print(type(reserve_suffisante))
````

### `=` et `==` : deux rôles distincts
`energie_wh = 20` est une **instruction d'affectation** : elle associe le nom à une valeur. `energie_wh == 20` est une **expression de comparaison** : elle vérifie une égalité sans modifier l'énergie.

Dans l'exemple précédent, Python évalue d'abord `energie_restante_wh >= reserve_minimale_wh`, puis affecte le booléen obtenu à `reserve_suffisante`. Les booléens s'écrivent sans guillemets et avec une majuscule ; `"True"` est du texte.

## 5. Conditions — choisir les instructions exécutées
### 5.1. `if` : agir seulement si le critère est vrai
`if` signifie « si ». Python évalue la condition ; si elle vaut `True`, il exécute le bloc associé. Sinon, il saute ce bloc et poursuit après lui.

```python
if condition:
    instruction_du_bloc
instruction_apres_le_bloc
```

Ce schéma décrit la syntaxe ; les noms génériques sont à remplacer. Le **deux-points `:`** annonce le bloc. L'**indentation** est le décalage en début de ligne : utilisez **quatre espaces**, de façon constante. Elle fait partie du programme, pas seulement de sa présentation. Le retour à l'alignement précédent marque la fin du bloc.

````python
energie_wh = 15
if energie_wh < 20:
    print("Réserve faible")
    print("Prévenir l'équipe")
print("Analyse terminée")
````

### 5.2. `if` / `else` : deux chemins complémentaires
`else` signifie « sinon ». Son bloc est exécuté quand la condition du `if` est fausse. `else` n'a **pas de condition écrite à sa suite**. Dans cette structure, exactement un des deux blocs est exécuté.

`if` et `else` sont alignés. Chaque bloc possède sa propre indentation ; les instructions après la structure retrouvent l'alignement de départ.

````python
energie_restante_wh = 0
if energie_restante_wh >= 0:
    print("Bilan énergétique non déficitaire")
else:
    print("Énergie insuffisante pour la mission")
print("Fin du bilan")
````

Ici, 0 Wh est classé dans le premier chemin : le bilan n'est pas déficitaire, mais il n'offre aucune réserve. **Le programme applique précisément le critère écrit** ; il ne déduit pas une marge de sécurité qui n'a pas été demandée.

### 5.3. `if` / `elif` / `else` : plusieurs cas ordonnés
`elif` signifie « sinon, si ». Python teste les conditions **de haut en bas** et exécute uniquement le bloc de la **première condition vraie**. Les conditions suivantes ne sont alors plus testées. Le `else` final couvre les cas restants ; il est facultatif.

Pour classer une réserve : en dessous de 0 Wh, la mission est déficitaire ; de 0 inclus à 20 exclus, la réserve est faible ; à partir de 20 Wh, la réserve est suffisante.

````python
energie_wh = 12
if energie_wh < 0:
    print("Déficit")
elif energie_wh < 20:
    print("Réserve faible")
else:
    print("Réserve suffisante")
````

Le `elif` ne répète pas `energie_wh >= 0` : lorsqu'on l'atteint, le premier test était faux. Cette information permet de comprendre la plage couverte.

## 6. Logique booléenne — combiner les critères
Les spécifications demandent souvent plusieurs conditions. Pour nos expressions booléennes, trois opérateurs suffisent :

| Écriture | Sens | Vraie lorsque… |
|---|---|---|
| `a and b` | a **et** b | les deux critères sont vrais |
| `a or b` | a **ou** b | au moins un critère est vrai, y compris les deux |
| `not a` | **non** a | le critère a est faux |

| `a` | `b` | `a and b` | `a or b` |
|---|---|---|---|
| `False` | `False` | `False` | `False` |
| `False` | `True` | `False` | `True` |
| `True` | `False` | `False` | `True` |
| `True` | `True` | `True` | `True` |

`not True` vaut `False` et `not False` vaut `True`. Nous appliquons ici `and`, `or` et `not` à des comparaisons ou à des variables booléennes. Évitez les critères implicites construits avec des nombres ou des textes.

````python
energie_wh = 30
liaison_active = True
arret_urgence = False
depart_autorise = (energie_wh >= 20) and liaison_active and (not arret_urgence)
print("Autorisation :", depart_autorise)
if depart_autorise:
    print("Départ")
else:
    print("Rester à la base")
````

Une température acceptable **de −10 °C à 45 °C, bornes incluses**, vérifie deux critères à la fois :

Python permet aussi la comparaison chaînée `-10 <= temperature_c <= 45`. Les deux écritures expriment ici le même intervalle. Être **hors** de cet intervalle signifie `temperature_c < -10 or temperature_c > 45` : un seul dépassement suffit.

Pour les expressions de ce chapitre, les comparaisons sont évaluées avant `not`, puis viennent `and`, puis `or`. Utilisez des parenthèses pour rendre votre intention lisible, surtout lorsque vous mélangez `and` et `or`.

## 7. Boucle `for` — répéter un bloc
Supposons que le rover doive envoyer trois messages de contrôle. Recopier trois fois `print(...)` fonctionne, mais oblige à réécrire le programme lorsque le nombre de contrôles change.

Une **boucle** répète un bloc. Une **itération** est un passage dans ce bloc. La boucle `for` attribue successivement les valeurs d'un parcours à une **variable d'itération**, puis exécute le bloc pour chacune. Nous utilisons `range()` pour fournir un parcours d'entiers.

```python
for etape in range(3):
    print("Contrôle", etape)
```

`for` se lit « pour chaque » ; `in` indique le parcours utilisé. `range(3)` fournit successivement **0, 1, 2**. Le deux-points et l'indentation délimitent le bloc, comme avec `if`. Le nom `etape` est choisi par nous ; `i` est courant, mais n'a aucun pouvoir particulier.

````python
for etape in range(3):
    print("Contrôle", etape)
    print("Message envoyé")
print("Séquence terminée")
````

### Exécution pas à pas
| Moment | Valeur attribuée à `etape` | Instructions exécutées |
|---|---:|---|
| Premier passage | 0 | les deux affichages indentés |
| Deuxième passage | 1 | les deux affichages indentés |
| Troisième passage | 2 | les deux affichages indentés |
| Après la boucle | — | « Séquence terminée » une seule fois |

La boucle termine tout son bloc avant de passer à la valeur suivante. Avec `range(3)`, elle effectue **trois** passages, même si la dernière valeur est 2. La répétition ne recommence pas le programme entier.

### Prédire — dans le bloc ou après ?

````python
# Ma prédiction, ligne par ligne :
for numero in range(3):
    print("Mesure", numero)
    print("Double", numero * 2)
print("Fin")
````

## 8. `range()` — choisir les entiers parcourus
`range()` décrit un parcours régulier d'entiers. Il ne lance pas la boucle à lui seul : `for` parcourt les valeurs qu'il fournit. On lui donne des entiers, pas une borne telle que `3.5`.

| Écriture | Départ | Arrêt exclu | Pas | Valeurs parcourues |
|---|---:|---:|---:|---|
| `range(4)` | 0 | 4 | 1 | 0, 1, 2, 3 |
| `range(2, 5)` | 2 | 5 | 1 | 2, 3, 4 |
| `range(2, 9, 3)` | 2 | 9 | 3 | 2, 5, 8 |
| `range(5, 0, -2)` | 5 | 0 | −2 | 5, 3, 1 |

**La borne d'arrêt est toujours exclue.** Avec un pas positif, on continue tant que la valeur reste strictement inférieure à cette borne. Avec un pas négatif, on continue tant qu'elle reste strictement supérieure. Le pas par défaut est 1 ; il ne peut pas être 0.

Pour parcourir les étapes **1 à 5 incluses**, écrivez `range(1, 6)`. Pour obtenir 0, 3, 6, 9, écrivez par exemple `range(0, 10, 3)` : la borne n'a pas besoin d'être atteinte.

## 9. Combiner `for` + `if` — examiner puis sélectionner
La boucle peut examiner chaque entier, tandis que la condition décide quoi faire **pour cet entier**. Le test est réévalué à chaque passage avec la nouvelle valeur de la variable d'itération.

Rappel : pour un entier `numero`, `numero % 3` donne le reste de la division par 3. `numero % 3 == 0` est donc vrai exactement lorsque `numero` est un multiple de 3.

### Exemple — transmettre tous les trois contrôles

````python
for numero in range(1, 8):
    if numero % 3 == 0:
        print("Transmission au contrôle", numero)
print("Parcours terminé")
````

### Deux niveaux de blocs
- Le `if` est décalé de **quatre espaces** : il appartient à la boucle.
- Le `print` de transmission est décalé de **huit espaces** : il appartient au `if`, lui-même dans la boucle.
- Le dernier `print` est à gauche : il est après la boucle.

La boucle effectue sept passages, mais affiche seulement deux transmissions, pour 3 et 6. **Nombre d'itérations et nombre d'actions sélectionnées sont différents.**

### Prédire — un message par étape
Écrivez les cinq lignes attendues. Le `else` est aligné avec quel `if` ?

````python
# Ma prédiction :
for etape in range(1, 6):
    if etape % 2 == 0:
        print(etape, "photographie")
    else:
        print(etape, "déplacement seul")
````

### 10.1. Compteur : combien de fois un critère est-il satisfait ?
Un **compteur** mémorise un nombre d'événements. On l'initialise à 0 **avant** la boucle, puis on ajoute 1 à chaque événement à compter.

Dans `nombre_photos = nombre_photos + 1`, Python lit l'ancienne valeur, calcule la nouvelle, puis la réaffecte au même nom. Ce n'est pas une égalité mathématique.

````python
nombre_photos = 0
for etape in range(1, 7):
    if etape % 2 == 0:
        nombre_photos = nombre_photos + 1
print("Photos :", nombre_photos)
````

### Suivre la mémoire pas à pas
| Étape | Étape paire ? | Compteur avant | Compteur après |
|---|---|---:|---:|
| 1 | non | 0 | 0 |
| 2 | oui | 0 | 1 |
| 3 | non | 1 | 1 |
| 4 | oui | 1 | 2 |
| 5 | non | 2 | 2 |
| 6 | oui | 2 | 3 |

Initialiser le compteur **dans** la boucle effacerait les événements déjà comptés. Ajouter 1 hors du `if` compterait tous les passages, au lieu des seules photos. Afficher après la boucle donne le bilan final ; afficher dans la boucle montrerait son évolution.

### 10.2. Accumulateur : quelle quantité totale ?
Un **accumulateur** conserve une somme construite au fil des passages. Le mécanisme est proche du compteur, mais on ajoute une **quantité**, éventuellement différente à chaque étape, au lieu d'ajouter systématiquement 1.

Dans cet exemple simplifié, l'étape 1 consomme 2 Wh, l'étape 2 consomme 4 Wh et l'étape 3 consomme 6 Wh. L'accumulateur part de 0 Wh.

````python
energie_totale_wh = 0
for etape in range(1, 4):
    energie_etape_wh = 2 * etape
    energie_totale_wh = energie_totale_wh + energie_etape_wh
    print("Étape", etape, "total provisoire :", energie_totale_wh, "Wh")
print("Consommation totale :", energie_totale_wh, "Wh")
````

Le total évolue de **0 à 2, puis 6, puis 12 Wh**. Écrire seulement `energie_totale_wh = energie_etape_wh` remplacerait le total par la dernière consommation : on perdrait les précédentes.

### 10.3. L'écriture abrégée `+=`
Pour les valeurs numériques de nos exemples :
```python
nombre_photos += 1
energie_totale_wh += energie_etape_wh
```
ces écritures réalisent respectivement `nombre_photos = nombre_photos + 1` et `energie_totale_wh = energie_totale_wh + energie_etape_wh`.

`+=` ne crée pas une valeur initiale : le nom doit déjà être défini. Attention à `=+ 1`, qui affecte simplement la valeur positive 1 ! Commencez avec l'écriture développée si elle rend le raisonnement plus clair.
