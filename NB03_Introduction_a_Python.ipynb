## 1. Une décision à chaque répétition
`for` répète un bloc pour chaque valeur du parcours. `if` exécute un bloc seulement si un critère est vrai.

**Comment prendre une décision à chaque répétition ?** Plaçons la condition **dans** la boucle. Nous ne changeons pas de langage : nous combinons deux outils connus.

Dans la suite, `range(1, 8)` fournit les entiers de 1 à 7 : la borne d'arrêt 8 est exclue. Pour un entier, `%` donne le reste d'une division ; `numero % 3 == 0` reconnaît un multiple de 3.

## 2. Combiner `for` + `if`
### Exemple repris de NB02 — transmettre tous les trois contrôles
Avant d'exécuter, prévoyez les numéros qui déclenchent une transmission.

```python
# Ma prédiction :
for numero in range(1, 8):
    if numero % 3 == 0:
        print("Transmission au contrôle", numero)
print("Parcours terminé")
```

### Deux niveaux de blocs : l'imbrication
Un bloc **imbriqué** est placé à l'intérieur d'un autre bloc.
- Le `if` est décalé de **quatre espaces** : il appartient au `for`.
- Le `print` de transmission est décalé de **huit espaces** : il appartient au `if`, lui-même dans le `for`.
- Le dernier `print` n'est pas décalé : il s'exécute **après** la boucle.

À chaque passage, Python attribue la prochaine valeur à `numero`, évalue le test, puis exécute ou saute le bloc du `if`. Il passe ensuite à l'itération suivante.

| `numero` | Multiple de 3 ? | Action du `if` |
|---|---|---|
| 1, puis 2 | non | aucun affichage |
| 3 | oui | transmission 3 |
| 4, puis 5 | non | aucun affichage |
| 6 | oui | transmission 6 |
| 7 | non | aucun affichage |

Il y a **7 itérations**, **2 transmissions**, puis **1 message final**. Un test faux ne met pas fin à la boucle.

### Prédire — trois positions pour un affichage
Écrivez les lignes attendues dans l'ordre. Combien de fois chacun des trois `print` s'exécute-t-il ?

```python
# Ma prédiction, ligne par ligne :
for numero in range(1, 4):
    if numero % 2 == 0:
        print("A : pair", numero)
    print("B : examiné", numero)
print("C : terminé")
```

## 3. Compteur — combien de multiples de 7 ?
Afficher les multiples de 7 entre 1 et 100 permet de les voir. Mais **comment obtenir leur nombre** sans compter les lignes à la main ?

Un **compteur** est une variable qui mémorise un nombre d'événements. Avant le parcours, aucun événement n'a été rencontré : on l'initialise à **0**. Chaque fois que le critère est vrai, on ajoute **1**.

Dans `compteur = compteur + 1`, Python lit la valeur actuelle à droite, ajoute 1, puis affecte le résultat au nom à gauche. Ce n'est pas une égalité mathématique : la valeur mémorisée change.

```python
compteur = 0
for numero in range(1, 101):
    if numero % 7 == 0:
        compteur = compteur + 1
print("Nombre de multiples de 7 :", compteur)
```

### Suivre l'état de la variable
| Passage | Compteur avant le test | Critère satisfait ? | Compteur après |
|---|---:|---|---:|
| 1 à 6, chaque passage | 0 | non | 0 |
| 7 | 0 | oui | 1 |
| 8 à 13, chaque passage | 1 | non | 1 |
| 14 | 1 | oui | 2 |
| … | … | … | … |
| 98 | 13 | oui | 14 |
| 99, puis 100 | 14 | non | 14 |

Après chaque passage, `compteur` représente **le nombre de multiples de 7 déjà rencontrés**. Il ne contient ni le dernier multiple, ni le nombre total de passages.

Pour les nombres, `compteur += 1` abrège `compteur = compteur + 1`. Il faut toujours définir la valeur initiale avant cette mise à jour.

**Trois emplacements, trois rôles :** initialiser **avant** la boucle ; augmenter **dans le `if`** ; afficher le bilan **après** la boucle. Initialiser à chaque passage effacerait le passé ; augmenter hors du `if` compterait tous les entiers.

## 4. Accumulateur — quelle quantité totale ?
Un **accumulateur** mémorise une somme construite progressivement. On ajoute une **quantité**, qui peut changer à chaque passage, au lieu d'ajouter toujours 1.

Commençons par la somme des entiers de 1 à 4. **Prédisez les totaux provisoires avant d'exécuter.**

```python
# Mes prédictions :
total = 0
for i in range(1, 5):
    total = total + i
    print("Après l'ajout de", i, ":", total)
print("Total final :", total)
```

| `i` | `total` avant | Calcul | `total` après |
|---:|---:|---|---:|
| 1 | 0 | 0 + 1 | 1 |
| 2 | 1 | 1 + 2 | 3 |
| 3 | 3 | 3 + 3 | 6 |
| 4 | 6 | 6 + 4 | 10 |

Après chaque passage, `total` contient **la somme des valeurs déjà parcourues**. `total = i` oublierait les valeurs précédentes. Pour ces nombres, `total += i` est l'écriture abrégée de `total = total + i`.

| Question | Mise à jour | Ce que représente la variable |
|---|---|---|
| Combien d'événements ? | `compteur += 1` | Un nombre d'événements |
| Quelle quantité cumulée ? | `total += quantite` | Une somme, avec l'unité de la quantité |

Attention : `total =+ i` affecte simplement la valeur positive de `i` ; ce n'est pas `+=`.

## 5. Pourquoi des collections ?
Pour trois températures, on pourrait écrire `temperature1`, `temperature2`, `temperature3`. Mais avec cent mesures, il faudrait cent noms et beaucoup de répétitions dans le code.

Une **liste** réunit plusieurs éléments dans un même objet, désigné par un seul nom. Elle est **ordonnée** : chaque élément a une position. Elle est **modifiable** : on peut remplacer un élément ou en ajouter. Deux éléments peuvent avoir la même valeur.

Les crochets `[...]` créent la liste ; les virgules séparent les éléments. Ici, toutes les valeurs sont des températures en °C et leur ordre représente celui des mesures.

```python
temperatures_c = [18, 21, 19, 24]
print(temperatures_c)
print("Nombre de mesures :", len(temperatures_c))
```

## 6. Lire et modifier une liste
### Les indices commencent à 0
Un **indice** est la position utilisée pour accéder à un élément. Avec le nom suivi de crochets, `temperatures_c[0]` lit le premier élément.

| Élément | 18 | 21 | 19 | 24 |
|---|---:|---:|---:|---:|
| Indice | 0 | 1 | 2 | 3 |

`len(temperatures_c)` vaut 4 : c'est le **nombre d'éléments**, pas le dernier indice. Pour une liste non vide, le dernier indice positif vaut `len(temperatures_c) - 1`. L'indice **−1** désigne directement le dernier élément.

### Prédire — position et valeur
Prévoyez les quatre résultats.

```python
# Mes prédictions :
temperatures_c = [18, 21, 19, 24]
print(temperatures_c[0])
print(temperatures_c[2])
print(temperatures_c[-1])
print(len(temperatures_c))
```

**Vérification :** 18, 19, 24, puis 4. La troisième mesure porte l'indice 2.

`temperatures_c[4]` provoquerait une `IndexError` : cette cinquième position n'existe pas. Une liste vide s'écrit `[]` et sa longueur vaut 0 ; elle n'a ni premier ni dernier élément.

### Remplacer ou ajouter : deux opérations différentes
- `temperatures_c[1] = 22` **remplace** la deuxième mesure ; la longueur ne change pas.
- `temperatures_c.append(26)` **ajoute** une mesure à la fin ; la longueur augmente de 1.

`append` est une **méthode** : une opération appelée sur la liste avec un point. Elle modifie la liste directement. N'écrivez pas `temperatures_c = temperatures_c.append(26)` : le résultat renvoyé par cette opération n'est pas la liste.

```python
temperatures_c = [18, 21, 19, 24]
temperatures_c[1] = 22
print("Après correction :", temperatures_c)
temperatures_c.append(26)
print("Après ajout :", temperatures_c)
print("Nombre de mesures :", len(temperatures_c))
```

## 7. Parcourir une liste avec `for`
`for` peut recevoir directement une liste. À chaque passage, la variable d'itération prend la **valeur de l'élément suivant**, dans l'ordre de la liste.

| Écriture | Valeurs prises par la variable |
|---|---|
| `for i in range(3):` | Les entiers 0, 1, 2 fournis par `range()` |
| `for temperature_c in [18, 21, 19]:` | Les températures 18, 21, 19 de la liste |

La variable `temperature_c` n'est donc **pas un indice**. Son nom au singulier aide à distinguer la mesure en cours de la collection `temperatures_c`.

**Prédisez** les affichages. Remplacez ensuite la deuxième mesure par 30 dans la liste et réexécutez : faut-il modifier la boucle ?

```python
# Ma prédiction :
temperatures_c = [18, 21, 19]
for temperature_c in temperatures_c:
    print("Mesure :", temperature_c, "°C")
print("Relevé terminé")
```

## 9. Premiers algorithmes sur listes
Un **algorithme** décrit les étapes qui permettent d'obtenir un résultat. Nous allons suivre ces étapes explicitement avant d'utiliser les outils de calcul déjà fournis par Python.

### Somme et moyenne à la main
Pour calculer une moyenne, on additionne toutes les mesures puis on divise par leur nombre. La division se fait **après** le parcours, quand la somme est complète.

**Prédisez** les valeurs successives de `total_c`, puis la moyenne.

```python
# Total après chaque passage et moyenne finale :
temperatures_c = [18, 22, 20]
total_c = 0
for temperature_c in temperatures_c:
    total_c += temperature_c
moyenne_c = total_c / len(temperatures_c)
print("Somme des mesures :", total_c, "°C")
print("Moyenne :", moyenne_c, "°C")
```

**Vérification :** le total passe de 0 à 18, puis 40, puis 60 ; la moyenne vaut 20 °C. La somme est une étape de calcul, pas la température du système. Ici la liste est **non vide** : avec `[]`, la division par sa longueur provoquerait une division par zéro.

### Maximum à la main : conserver le meilleur candidat
On initialise le maximum avec **le premier élément**, puis on compare chaque valeur au maximum mémorisé. Si elle est plus grande, elle devient le nouveau maximum.

Pourquoi ne pas partir de 0 ? Avec des températures toutes négatives, 0 ne serait même pas une mesure de la liste.

**Prédisez** le maximum mémorisé après chaque passage.

```python
# Mes prédictions :
temperatures_c = [-8, -3, -6]
maximum_c = temperatures_c[0]
for temperature_c in temperatures_c:
    if temperature_c > maximum_c:
        maximum_c = temperature_c
    print("Maximum provisoire :", maximum_c, "°C")
print("Maximum final :", maximum_c, "°C")
```

**Vérification :** −8, puis −3, puis −3. Comparer le premier élément à lui-même ne change rien et permet de parcourir simplement toute la liste. Après chaque passage, `maximum_c` est le maximum **des valeurs déjà examinées**.

### Après les algorithmes : les outils fournis par Python
Vous avez maintenant programmé le mécanisme. Python propose aussi `sum(liste)`, `min(liste)` et `max(liste)` pour obtenir la somme, le minimum et le maximum. Ces outils, comme `len()`, sont déjà disponibles : rien à installer.

Ils sont utiles pour vérifier vos résultats. Dans les exercices algorithmiques de ce notebook, conservez d'abord vos boucles pour montrer votre raisonnement.

```python
temperatures_c = [16, 20, 24]
print("Somme :", sum(temperatures_c))
print("Minimum :", min(temperatures_c))
print("Maximum :", max(temperatures_c))
print("Moyenne :", sum(temperatures_c) / len(temperatures_c))
```

`sum([])` vaut 0, mais `min([])` et `max([])` ne peuvent pas choisir un élément dans une liste vide et provoquent une `ValueError`. La moyenne d'une liste vide n'est pas définie ici. Nous supposons des listes non vides dans le socle.
